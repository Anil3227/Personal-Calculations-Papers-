```python
from getdist import MCSamples, plots
from matplotlib.patches import Patch
import matplotlib.pyplot as plt
import numpy as np
import matplotlib as mpl
mpl.rcParams['mathtext.fontset'] = 'cm'

# ---------- load chains ----------
data1 = np.loadtxt('/Users/anilkandel/Documents/SimpleMC/simplemc/Anil 2/wCDM_phy_PLK18+DESIDR2+Desdovekie_mcmc_1.txt')
data2 = np.loadtxt('/Users/anilkandel/Documents/SimpleMC/simplemc/Anil 2/wCDM_phy_PLK18+DESIDR2+DESY5_mcmc_1.txt')
data3 = np.loadtxt('/Users/anilkandel/Documents/SimpleMC/simplemc/Anil 2/wCDM_phy_DESIDR2+Planck_18+DESY5exc_mcmc_1.txt')

weights1 = np.ones(data1.shape[0])
weights2 = np.ones(data2.shape[0])
weights3 = np.ones(data3.shape[0])

# first 6 columns; param3 = w0
params1 = data1[:, :6]
params2 = data2[:, :6]
params3 = data3[:, :6]

param_names  = ['param1','param2','Om','Obh2','h','param3']
param_labels = ['Param1','Param2',r'\Omega_m',r'\Omega_b h^2',r'h',r'w']

samples1 = MCSamples(samples=params1, weights=weights1, names=param_names, labels=param_labels)
samples2 = MCSamples(samples=params2, weights=weights2, names=param_names, labels=param_labels)
samples3 = MCSamples(samples=params3, weights=weights3, names=param_names, labels=param_labels)

# ------------------- 2D Plot for Om and w0 -------------------
g2d = plots.get_single_plotter(width_inch=6)
g2d.settings.figure_legend_frame = True
g2d.settings.alpha_filled_add = 0.7
g2d.settings.axes_labelsize = 25
g2d.settings.legend_fontsize = 10

# Create a 2D plot for Om and w0 for both chains
g2d.plot_2d(
    [samples1, samples2, samples3],  # List of MCSamples objects    
    param_pair=['param3', 'Om'],  # w0 and Om
    filled=[True, True, True],
    colors=['#0B1291', 'magenta', 'red'],   # dark blue for DESDovekie, magenta for DESY5, red for Planck
    contour_ls=['-', '-', '-'],           # dashed for dark blue, solid for magenta, dotted for red
)

# ------------------- Force dashed line for first contour -------------------
ax = g2d.subplots[0,0] # type: ignore

# ------------------- Add legend with dashed line -------------------
handle1 = Patch(facecolor='darkblue', edgecolor='darkblue', label=r'CMB+DESI DR2+DES-Dovekie')
handle2 = Patch(facecolor='magenta', edgecolor='magenta', label=r'CMB+DESI DR2+DESY5')
handle3 = Patch(facecolor='red', edgecolor='red', label=r'CMB+DESI DR2+DESY5 ($z>0.1$)')
# Add legend to the axis
ax.legend(handles=[handle1, handle2, handle3], fontsize=9, loc='upper right',
          bbox_to_anchor=(0.52, 1.01), frameon=True)
# ------------------- Add reference lines and ΛCDM point -------------------
ax = g2d.subplots[0, 0]  # type: ignore # Get the axis from GetDist plotter


# Vertical and horizontal dashed lines
ax.axvline(x=-1.0, color='gray', linestyle='--', linewidth=1.2)
ax.text(
    -0.99,                   # small shift to the right
    ax.get_ylim()[1]*0.83,
    r'$\Lambda$CDM',
    rotation=90,
    color='black',
    fontsize=13
)
# Export the 2D plot
g2d.export("pdfs/wCDM.pdf")
```

```python
from getdist import MCSamples, plots
import matplotlib.pyplot as plt
import numpy as np
import matplotlib as mpl
mpl.rcParams['mathtext.fontset'] = 'cm'


# ---------- load chains ----------
data1 = np.loadtxt('/Users/anilkandel/Documents/SimpleMC/simplemc/Anil 2/CPL_phy_PLK18+DESIDR2+Desdovekie_mcmc_1.txt')
data2 = np.loadtxt('/Users/anilkandel/Documents/SimpleMC/simplemc/Anil 2/CPL_phy_PLK18+DESIDR2+DESY5_mcmc_1.txt')
data3 = np.loadtxt('/Users/anilkandel/Documents/SimpleMC/simplemc/Anil 2/CPL_phy_DESIDR2+Planck_18+DESY5exc_mcmc_1.txt')



weights1 = np.ones(data1.shape[0])
weights2 = np.ones(data2.shape[0])
weights3 = np.ones(data3.shape[0])

# first 7 columns; param3 = w0, param5 = wa

params1 = data1[:, :7]
params2 = data2[:, :7]
params3 = data3[:, :7]

param_names  = ['param1','param2','Om','Obh2','h','param3','param5']
param_labels = ['Param1','Param2',r'\Omega_m',r'\Omega_b h^2',r'h',r'w_0',r'w_a']

samples1 = MCSamples(samples=params1, weights=weights1, names=param_names, labels=param_labels)
samples2 = MCSamples(samples=params2, weights=weights2, names=param_names, labels=param_labels)
samples3 = MCSamples(samples=params3, weights=weights3, names=param_names, labels=param_labels)

# ------------------- 2D Plot for w0 and wa -------------------
g2d = plots.get_single_plotter(width_inch=6)
g2d.settings.figure_legend_frame = True
g2d.settings.alpha_filled_add = 0.7
g2d.settings.axes_labelsize = 25
g2d.settings.legend_fontsize = 10

# Create a 2D plot for w0 and wa for both chains
g2d.plot_2d(
    [samples1, samples2, samples3],
    param_pair=['param3', 'param5'],  # w0 and wa
    filled=[True, True, True],
    colors=['#0B1291', 'magenta', 'red'],   # dark blue for DESDovekie, magenta for DESY5, red for Planck
    contour_ls=['-', '-', '-'],           # dashed for dark blue, solid for magenta, dotted for red
)
# ------------------- Force dashed line for first contour -------------------
ax = g2d.subplots[0,0] # type: ignore


# ------------------- Add legend with dashed line -------------------
# Create custom legend handles
handle1 = Patch(facecolor='darkblue', edgecolor='#0B1291', label=r'CMB+DESI DR2+DES-Dovekie')
handle2 = Patch(facecolor='magenta', edgecolor='magenta', label=r'CMB+DESI DR2+DESY5')
handle3 = Patch(facecolor='red', edgecolor='red', label=r'CMB+DESI DR2+DESY5 ($z>0.1$)')

# Add legend to the axis
ax.legend(handles=[handle1, handle2, handle3], fontsize=9, loc='upper right',
          bbox_to_anchor=(1.01, 1.01), frameon=True)
# ------------------- Add reference lines and ΛCDM point -------------------
ax = g2d.subplots[0, 0]  # type: ignore # Get the axis from GetDist plotter

# Vertical and horizontal dashed lines
ax.axvline(x=-1, color='gray', linestyle='--', linewidth=1.2)
ax.axhline(y=0, color='gray', linestyle='--', linewidth=1.2)

# ΛCDM marker and annotation
ax.plot([-1], [0], marker='*', ms=12, color='k', zorder=5)

ax.annotate(
    r'$\Lambda$CDM',
    xy=(-1, 0),
    xytext=(-0.98, 0.45),
    fontsize=11,
    fontweight='bold',
    color='black',
    arrowprops=dict(arrowstyle='->', lw=1.2, color='black'),
    zorder=5
)

# Export the 2D plot
g2d.export("pdfs/CPL.pdf")
```
